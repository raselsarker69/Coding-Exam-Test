```python
def sum_even_number(num):
  sum_even = 0
  n = 2

  while n <= num:
    if n % 2 == 0:
      sum_even += n
    n += 2
  return sum_even

num = int(input("Enter your number:"))
result = sum_even_number(num)
print(result)
```

# prob: 01 Gradient Computation for Linear Regression

```python
def gradient_compute(weight, bias):
  weight_grad = 2 * weight
  bias_grad = 1
  return weight_grad, bias_grad

x = torch.tensor([1.0, 2.0, 3.0])
y = torch.tensor([2.0, 4.0, 6.0])

w = torch.tensor(0.0, requires_grad=True) #Enable gradient tracking
b = torch.tensor(0.0, requires_grad=True)

y_pred = w * x + b
loss = torch.mean((y_pred - y) ** 2)
loss.backward() #backward pass

w_grad = w.grad.item()
b_grad = b.grad.item()

#Access gradients
print("Gradients for w and b: ", w_grad, b_grad)
```

# Prob:02 Hessian Approximation for a Quadratic Function

```python
import torch

def quadratic_function(w, A, b, c):
    return w @ A @ w + b @ w + c

def approximate_hessian(f, w, A, b, c, epsilon=1e-4):
    n = w.numel()
    hessian = torch.zeros((n, n))

    for i in range(n):
        # +epsilon perturbation
        w_i_plus = w.clone().detach()
        w_i_plus[i] = w_i_plus[i] + epsilon
        w_i_plus.requires_grad_(True)
        grad_plus = torch.autograd.grad(f(w_i_plus, A, b, c), w_i_plus)[0]

        # -epsilon perturbation
        w_i_minus = w.clone().detach()
        w_i_minus[i] = w_i_minus[i] - epsilon
        w_i_minus.requires_grad_(True)
        grad_minus = torch.autograd.grad(f(w_i_minus, A, b, c), w_i_minus)[0]

        # central diff
        hessian[:, i] = (grad_plus - grad_minus) / (2 * epsilon)

    return hessian


# input
A = torch.tensor([[2.0, 1.0], [1.0, 2.0]])
b = torch.tensor([1.0, 1.0])
c = 0.0
w = torch.tensor([1.0, 1.0])

# Compute Hessian
H = approximate_hessian(quadratic_function, w, A, b, c)
print("Approximate Hessian:\n", H)
```

Prob: 05 Tracking NaN Loss in Neural Network Training

```python
import torch
import torch.nn as nn
import torch.optim as optim

#============ problem understand===========================#
# Use SGD optimizer
# Use MSE loss
# Up to 100 epochs
# Learning rate (e.g., 10.0) might cause instability → NaN
# If NaN occurs, return epoch number
# else, return -1
#===========================================================#

def detect_nan_loss(x, y):
    # define 2-layer MLP
    model = nn.Sequential(
        nn.Linear(10, 50),
        nn.ReLU(),
        nn.Linear(50, 1)
    )

    # Loss and optimizer
    criterion = nn.MSELoss()
    optimizer = optim.SGD(model.parameters(), lr=10.0)

    # training loop with 100 epoch
    for epoch in range(100):
        model.train()
        optimizer.zero_grad()
        output = model(x)
        loss = criterion(output, y)

        # Check nan loss
        if torch.isnan(loss):
            return epoch  # return where NaN occurred

        loss.backward()
        optimizer.step()

    return -1  # if no NaN occurred

# input
x = torch.randn(100, 10)
y = torch.randn(100, 1)

nan_epoch = detect_nan_loss(x, y)
print("NaN detected at epoch:" if nan_epoch != -1 else "No NaN found", nan_epoch)
```
