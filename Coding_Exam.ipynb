```python
def reverse_char(s):
  s_reverse = " "
  for chr in s:
    s_reverse = chr + s_reverse
  return s_reverse


st = input("Enter your string: ")
result = reverse_char(st)
print(result)
```

```python
def sum_even_number(num):
  sum_even = 0
  n = 2

  while n <= num:
    if n % 2 == 0:
      sum_even += n
    n += 2
  return sum_even

num = int(input("Enter your number:"))
result = sum_even_number(num)
print(result)
```

```python
def factorial_calculete(num):
  if num == 0:
    return 1
  else:
    return num * factorial_calculete(num - 1)

num = int(input("Enter your number:"))
result = factorial_calculete(num)
print(result)
```

```python
def marge_sorted_list(list1, list2):
  marge_list = list1 + list2
  marge_list.sort()
  return marge_list

#input
list1 = [5,7,10, 35, 90]
list2 = [20,40, 45, 80, 100]

result = marge_sorted_list(list1, list2)
print(result)
```
